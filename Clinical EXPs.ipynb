```python
#EXP2

import pandas as pd
import numpy as np

# Import dataset
df = pd.read_csv("clinical_trial_data_raw.csv")

print("Original Dataset Shape:", df.shape)

# 1. Check missing values
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

# 2. Check duplicate rows
print("\nExact Duplicate Rows:", df.duplicated().sum())

# 3. Check duplicate patient IDs
print("\nDuplicate Patient IDs:", df.duplicated(subset=['USUBJID']).sum())

# 4. Check inconsistent categorical values
print("\nSEX Values Before Cleaning:")
print(df['SEX'].value_counts(dropna=False))

print("\nARM Values Before Cleaning:")
print(df['ARM'].value_counts(dropna=False))

# 5. Check invalid ages
print("\nInvalid AGE Values:")
print(df[~df['AGE'].between(18, 100)]['AGE'])

# 6. Remove exact duplicate rows
before = len(df)
df = df.drop_duplicates()
print("\nExact Duplicate Rows Removed:", before - len(df))

# 7. Remove duplicate patient IDs (keep first occurrence)
before = len(df)
df = df.drop_duplicates(subset=['USUBJID'], keep='first')
print("Duplicate Patient IDs Removed:", before - len(df))

# 8. Standardize SEX values
df['SEX'] = df['SEX'].astype('string').str.strip().str.upper()

df['SEX'] = df['SEX'].replace({
    'MALE': 'M',
    'FEMALE': 'F'
})

# 9. Standardize ARM values
df['ARM'] = df['ARM'].astype('string').str.strip().str.lower()

df['ARM'] = df['ARM'].replace({
    'drug10mg': 'Drug 10mg',
    'placebo': 'Placebo'
})

# 10. Handle invalid AGE values
df.loc[~df['AGE'].between(18, 100), 'AGE'] = np.nan

# 11. Standardize date formats
df['ENROLL_DATE'] = pd.to_datetime(
    df['ENROLL_DATE'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
).dt.strftime('%Y-%m-%d')

# 12. Handle missing values
numeric_columns = [
    'AGE', 'WEIGHT_KG', 'BMI',
    'GLUCOSE_MGDL', 'CHOLESTEROL_MGDL',
    'HEARTRATE'
]

for col in numeric_columns:
    if col in df.columns and df[col].isnull().sum() > 0:
        count = df[col].isnull().sum()
        median_value = df[col].median()
        df[col] = df[col].fillna(median_value)
        print(f"{col}: {count} missing values filled with median ({median_value})")

# Fill missing categorical SITEID values
if 'SITEID' in df.columns:
    count = df['SITEID'].isnull().sum()
    df['SITEID'] = df['SITEID'].fillna('Unknown')
    print(f"SITEID: {count} missing values filled with Unknown")

# 13. Check missing values after cleaning
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# 14. Check duplicate records after cleaning
print("\nDuplicate Rows After Cleaning:", df.duplicated().sum())
print("Duplicate Patient IDs After Cleaning:",
      df.duplicated(subset=['USUBJID']).sum())

# 15. Display cleaned dataset
print("\nCleaned Dataset Shape:", df.shape)
display(df.head())

# 16. Save cleaned dataset
df.to_csv("clinical_trial_data_clean.csv", index=False)

print("\nCleaned Dataset Saved Successfully!")
print("Experiment 2 Completed Successfully!")
```

```python
#EXP3

import pandas as pd
import numpy as np

# Import datasets
demographics = pd.read_csv("clinical_demographics.csv")
vitals_labs = pd.read_csv("clinical_vitals_labs.csv")

# Display dataset shapes
print("Demographics Shape:", demographics.shape)
print("Vitals and Labs Shape:", vitals_labs.shape)

# Check uniqueness of merge keys
print("\nDuplicate IDs in Demographics:",
      demographics['USUBJID'].duplicated().sum())

print("Duplicate IDs in Vitals and Labs:",
      vitals_labs['USUBJID'].duplicated().sum())

# Check unmatched patient IDs
print("\nIDs present only in Demographics:")
print(set(demographics['USUBJID']) - set(vitals_labs['USUBJID']))

print("\nIDs present only in Vitals and Labs:")
print(set(vitals_labs['USUBJID']) - set(demographics['USUBJID']))

# Merge datasets
df = pd.merge(
    demographics,
    vitals_labs,
    on='USUBJID',
    how='inner',
    validate='one_to_one'
)

print("\nMerged Dataset Shape:", df.shape)
print("Expected Number of Rows: 180")

# Create AGE_GROUP
df['AGE_GROUP'] = pd.cut(
    df['AGE'],
    bins=[0, 39, 59, np.inf],
    labels=['18-39', '40-59', '60+']
)

# Create BMI_CATEGORY
df['BMI_CATEGORY'] = pd.cut(
    df['BMI'],
    bins=[-np.inf, 18.5, 25, 30, np.inf],
    labels=['Underweight', 'Normal', 'Overweight', 'Obese'],
    right=False
)

# Create BP_CATEGORY
conditions = [
    (df['SYSBP'] < 120) & (df['DIABP'] < 80),
    (df['SYSBP'] < 130) & (df['DIABP'] < 80),
    (df['SYSBP'] >= 130) | (df['DIABP'] >= 80)
]

choices = ['Normal', 'Elevated', 'High']

df['BP_CATEGORY'] = np.select(
    conditions,
    choices,
    default='Unknown'
)

# Create DAYS_ON_STUDY
df['ENROLL_DATE'] = pd.to_datetime(
    df['ENROLL_DATE'],
    errors='coerce'
)

reference_date = pd.Timestamp('2025-12-31')

df['DAYS_ON_STUDY'] = (
    reference_date - df['ENROLL_DATE']
).dt.days

# Create HIGH_RISK_FLAG
df['HIGH_RISK_FLAG'] = np.where(
    (df['SMOKER'].astype(str).str.strip().str.upper().isin(['YES', 'Y', '1', 'TRUE'])) |
    (df['CHOLESTEROL_MGDL'] >= 240) |
    (df['GLUCOSE_MGDL'] >= 126),
    'Yes',
    'No'
)

# Display derived variable summaries
print("\nAGE GROUP COUNTS:")
print(df['AGE_GROUP'].value_counts(dropna=False))

print("\nBMI CATEGORY COUNTS:")
print(df['BMI_CATEGORY'].value_counts(dropna=False))

print("\nBP CATEGORY COUNTS:")
print(df['BP_CATEGORY'].value_counts(dropna=False))

print("\nDAYS ON STUDY SUMMARY:")
print(df['DAYS_ON_STUDY'].describe())

print("\nHIGH RISK FLAG COUNTS:")
print(df['HIGH_RISK_FLAG'].value_counts(dropna=False))

# Display transformed dataset
print("\nTransformed Dataset:")
display(df.head())

# Export final dataset
df.to_csv("clinical_trial_data_analysis.csv", index=False)

print("\nMerged and Transformed Dataset Saved Successfully!")
print("Final Dataset Shape:", df.shape)
print("Experiment 3 Completed Successfully!")
```

```python
#EXP5

import pandas as pd
import numpy as np

dm_raw = pd.read_csv('sdtm_raw_demographics.csv')
vs_raw = pd.read_csv('sdtm_raw_vitals.csv')
lb_raw = pd.read_csv('sdtm_raw_labs.csv')
ae_raw = pd.read_csv('sdtm_raw_adverse_events.csv')
ex_raw = pd.read_csv('sdtm_raw_exposure.csv')

# DM Domain
dm = dm_raw.copy()
dm['STUDYID'] = 'CT01'
dm['DOMAIN'] = 'DM'
dm['DMSEQ'] = 1
dm['ARMCD'] = dm['ARM'].str.split().str[0] + dm['ARM'].str.extract(r'(\d+)')[0]
dm['AGEUNIT'] = 'YEARS'
dm['SEXCD'] = dm['SEX'].map({'M': 'M', 'F': 'F'})
dm['RACECD'] = dm['RACE']

dm_sdtm = dm[['STUDYID', 'DOMAIN', 'USUBJID', 'DMSEQ', 'AGE', 'AGEUNIT',
              'SEXCD', 'RACECD', 'ARM', 'ARMCD', 'RFSTDTC', 'RFENDTC']]

dm_sdtm.to_csv('dm_sdtm.csv', index=False)

# VS Domain
vs = vs_raw.melt(
    id_vars=['USUBJID', 'VISIT', 'VISITNUM', 'VISITDATE'],
    value_vars=['SYSBP', 'DIABP', 'HEARTRATE'],
    var_name='VSTESTCD',
    value_name='VSRESN'
)

vs['STUDYID'] = 'CT01'
vs['DOMAIN'] = 'VS'
vs['VSTEST'] = vs['VSTESTCD'].map({
    'SYSBP': 'Systolic Blood Pressure',
    'DIABP': 'Diastolic Blood Pressure',
    'HEARTRATE': 'Heart Rate'
})
vs['VSUNIT'] = vs['VSTESTCD'].map({
    'SYSBP': 'mmHg',
    'DIABP': 'mmHg',
    'HEARTRATE': 'bpm'
})
vs['VSSTDTC'] = vs['VISITDATE']
vs['VSSEQ'] = vs.groupby('USUBJID').cumcount() + 1

vs_sdtm = vs[['STUDYID', 'DOMAIN', 'USUBJID', 'VSSEQ', 'VISIT',
              'VISITNUM', 'VSSTDTC', 'VSTEST', 'VSTESTCD', 'VSRESN', 'VSUNIT']]

vs_sdtm.to_csv('vs_sdtm.csv', index=False)

# LB Domain
lb = lb_raw.copy()
lb['STUDYID'] = 'CT01'
lb['DOMAIN'] = 'LB'
lb['LBSTDTC'] = lb['VISITDATE']
lb['LBTESTCD'] = lb['LBTEST'].str.replace(' ', '_').str.upper()
lb['LBSEQ'] = lb.groupby('USUBJID').cumcount() + 1
lb['LBLOINC'] = '2345-7'

lb_sdtm = lb[['STUDYID', 'DOMAIN', 'USUBJID', 'LBSEQ', 'VISIT',
              'LBSTDTC', 'LBTEST', 'LBTESTCD', 'LBRESN', 'LBUNIT', 'LBREFRNC']]

lb_sdtm.to_csv('lb_sdtm.csv', index=False)

# AE Domain
if len(ae_raw) > 0:
    ae = ae_raw.copy()
    ae['STUDYID'] = 'CT01'
    ae['DOMAIN'] = 'AE'
    ae['AESEQ'] = ae.groupby('USUBJID').cumcount() + 1
    ae['AESEVCD'] = ae['AESEV'].str[0]
    ae['AERELCD'] = ae['AEREL'].str[0]

    ae_sdtm = ae[['STUDYID', 'DOMAIN', 'USUBJID', 'AESEQ',
                  'AETERM', 'AESTDTC', 'AEENDTC', 'AESEV', 'AEREL', 'AEOUT']]

    ae_sdtm.to_csv('ae_sdtm.csv', index=False)
else:
    ae_sdtm = pd.DataFrame()
    print('No adverse event records found.')

# EX Domain
ex = ex_raw.copy()
ex['STUDYID'] = 'CT01'
ex['DOMAIN'] = 'EX'
ex['EXSEQ'] = ex.groupby('USUBJID').cumcount() + 1
ex['EXDTC'] = ex['EXSTDTC']

ex_sdtm = ex[['STUDYID', 'DOMAIN', 'USUBJID', 'EXSEQ',
              'EXDRUG', 'EXDOSE', 'EXUNIT', 'EXSTDTC',
              'EXENDTC', 'EXFREQ', 'EXROUTE']]

ex_sdtm.to_csv('ex_sdtm.csv', index=False)

# Validation
domains = {
    'DM': dm_sdtm,
    'VS': vs_sdtm,
    'LB': lb_sdtm,
    'AE': ae_sdtm,
    'EX': ex_sdtm
}

for name, data in domains.items():
    if not data.empty:
        print(f'\n{name} Domain')
        print('Records:', len(data))
        print('Missing values:', data.isnull().sum().sum())
        print('Duplicate records:', data.duplicated().sum())
        print('Missing USUBJID:', data['USUBJID'].isnull().sum())
        print('Unique subjects:', data['USUBJID'].nunique())

print('\nAll SDTM domains created successfully!')
print('Files saved: dm_sdtm.csv, vs_sdtm.csv, lb_sdtm.csv, ae_sdtm.csv, ex_sdtm.csv')
```

```python
#EXP6
import pandas as pd
import numpy as np

# Load SDTM datasets
dm = pd.read_csv('adam_input_dm_sdtm.csv')
vs = pd.read_csv('adam_input_vs_sdtm.csv')
lb = pd.read_csv('adam_input_lb_sdtm.csv')
ae = pd.read_csv('adam_input_ae_sdtm.csv')
ex = pd.read_csv('adam_input_ex_sdtm.csv')

# ADSL - Subject Level
base_sbp = vs[vs['VSTESTCD'].isin(['SYSBP'])].sort_values('VISITNUM').groupby('USUBJID')['VSRESN'].first()

adsl = dm.copy()
adsl['BASE_SBP'] = adsl['USUBJID'].map(base_sbp)
adsl['AGROUP'] = pd.cut(adsl['AGE'], bins=[0, 40, 60, 150],
                        labels=['18-40', '41-60', '60+'])
adsl['HIGH_RISK'] = np.where((adsl['AGE'] >= 60) & (adsl['BASE_SBP'] >= 140), 'Yes', 'No')
adsl['SAFFL'] = 'Y'

dose = ex.groupby('USUBJID')['EXDOSE'].first().reset_index()
adsl = adsl.merge(dose, on='USUBJID', how='left')
adsl.to_csv('adam_adsl.csv', index=False)

# Function to create ADVS and ADLB
def create_analysis(df, test_col, value_col, output, lab=False):
    df = df.copy()
    df = df.sort_values('VISITNUM')

    df['BASE'] = df.groupby(['USUBJID', test_col])[value_col].transform(
        lambda x: x.iloc[0]
    )
    df['AVAL'] = df[value_col]
    df['CHG'] = df['AVAL'] - df['BASE']
    df['PCHG'] = np.where(df['BASE'] != 0, (df['CHG'] / df['BASE']) * 100, np.nan)

    if lab:
        df['RESPONSE'] = np.where(df['CHG'] <= -10, 'Y', 'N')
    else:
        df['RESPONSE'] = np.where(
            ((df[test_col] == 'SYSBP') & (df['CHG'] <= -10)) |
            ((df[test_col] == 'DIABP') & (df['CHG'] <= -5)), 'Y', 'N'
        )

    df.to_csv(output, index=False)
    return df

# ADVS - Vital Signs
advs = create_analysis(vs, 'VSTESTCD', 'VSRESN', 'adam_advs.csv')

# ADLB - Laboratory
adlb = create_analysis(lb, 'LBTESTCD', 'LBRESN', 'adam_adlb.csv', lab=True)

# ADAE - Adverse Events
adae = ae.merge(
    adsl[['USUBJID', 'ARM', 'ARMCD', 'EXDOSE']],
    on='USUBJID', how='left'
)

adae['AESEVN'] = adae['AESEV'].map({'Mild': 1, 'Moderate': 2, 'Severe': 3})
adae['AERELN'] = adae['AEREL'].map({'Related': 1, 'Unrelated': 0})
adae['AESEQ'] = adae.groupby('USUBJID').cumcount() + 1

adae.to_csv('adam_adae.csv', index=False)

# Validation and summary report
datasets = {
    'ADSL': adsl,
    'ADVS': advs,
    'ADLB': adlb,
    'ADAE': adae
}

for name, df in datasets.items():
    key = ['USUBJID'] if name == 'ADSL' else (
        ['USUBJID', 'AESEQ'] if name == 'ADAE' else
        ['USUBJID', 'VISITNUM', 'VSTESTCD' if name == 'ADVS' else 'LBTESTCD']
    )

    print(f'\n{name}')
    print('Records:', len(df))
    print('Missing values:', df.isnull().sum().sum())
    print('Duplicate keys:', df.duplicated(subset=key).sum())

print('\nADaM datasets created successfully!')
print('Files: adam_adsl.csv, adam_advs.csv, adam_adlb.csv, adam_adae.csv')
```

```python
#EXP7
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Load and inspect data
a = pd.read_csv('adam_exp7_analysis.csv')
print('Shape:', a.shape)
print(a.dtypes)
print('\nMissing values:\n', a.isnull().sum())

# Summary statistics
for col in ['SBP_CHG', 'DBP_CHG', 'GLU_CHG']:
    print(f'\n{col} Summary:')
    print(a.groupby('ARM')[col].agg(['count', 'mean', 'std']).round(2))

# Normality plots for Placebo SBP
pbo = a.loc[a['ARM'] == 'Placebo', 'SBP_CHG'].dropna()

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.hist(pbo, bins=10, edgecolor='black')
plt.title('Placebo SBP Change - Histogram')

plt.subplot(1, 2, 2)
stats.probplot(pbo, dist='norm', plot=plt)
plt.title('Q-Q Plot')
plt.tight_layout()
plt.show()

# T-test, confidence interval and Cohen's d for all endpoints
for col in ['SBP_CHG', 'DBP_CHG', 'GLU_CHG']:
    print(f'\n========== {col} ==========')

    pbo = a.loc[a.ARM == 'Placebo', col].dropna()
    d20 = a.loc[a.ARM == 'Drug 20mg', col].dropna()

    lev = stats.levene(pbo, d20)
    equal_var = lev.pvalue >= 0.05

    t, p = stats.ttest_ind(d20, pbo, equal_var=equal_var)
    diff = d20.mean() - pbo.mean()

    n1, n2 = len(d20), len(pbo)
    sp = np.sqrt(((n1-1)*d20.var() + (n2-1)*pbo.var()) / (n1+n2-2))
    d = diff / sp

    if equal_var:
        df = n1 + n2 - 2
        se = sp * np.sqrt(1/n1 + 1/n2)
    else:
        se = np.sqrt(d20.var()/n1 + pbo.var()/n2)
        df = (d20.var()/n1 + pbo.var()/n2)**2 / (
            (d20.var()/n1)**2/(n1-1) + (pbo.var()/n2)**2/(n2-1)
        )

    margin = stats.t.ppf(0.975, df) * se
    print(f'Levene p={lev.pvalue:.4f}')
    print(f't={t:.3f}, df={df:.2f}, p={p:.4f}')
    print(f'Mean difference={diff:.2f}')
    print(f'95% CI=({diff-margin:.2f}, {diff+margin:.2f})')
    print(f"Cohen's d={d:.3f}")
    print('Conclusion:', 'Significant difference' if p < 0.05 else 'No significant difference')

    # One-way ANOVA
    groups = [g[col].dropna() for _, g in a.groupby('ARM')]
    f, pa = stats.f_oneway(*groups)
    print(f'ANOVA: F={f:.3f}, p={pa:.4f}')

    if pa < 0.05:
        print('Tukey HSD:')
        print(pairwise_tukeyhsd(a[col].dropna(), a.loc[a[col].notna(), 'ARM']))
    else:
        print('Post-hoc test not required.')

# Chi-square test for adverse events
print('\n========== ADVERSE EVENTS ==========')
ct = pd.crosstab(a['ARM'], a['AE_YN'])
print('Counts:\n', ct)
print('Percentages:\n', (ct.div(ct.sum(axis=1), axis=0)*100).round(2))

chi2, p, dof, expected = stats.chi2_contingency(ct)

if (expected < 5).any() and ct.shape == (2, 2):
    odds, p = stats.fisher_exact(ct)
    print(f"Fisher Exact Test: p={p:.4f}")
else:
    print(f'Chi-square={chi2:.3f}, df={dof}, p={p:.4f}')

print('Conclusion:', 'Significant association' if p < 0.05 else 'No significant association')

print('\nExperiment 7 completed successfully!')
```

```python
#EXP 8
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Load datasets
a = pd.read_csv('adam_exp7_analysis.csv')
dm = pd.read_csv('adam_input_dm_sdtm.csv')
ae = pd.read_csv('adam_input_ae_sdtm.csv')

# Analysis population
a = a[a['SAFFL'] == 'Y'].merge(dm[['USUBJID', 'RACECD']], on='USUBJID', how='left')
arms = ['Placebo', 'Drug 10mg', 'Drug 20mg']
N = {k: (a.ARM == k).sum() for k in arms}
print('Analysis population:', len(a))

def fmt(n, total):
    return f'{n} ({100*n/total:.1f}%)' if total else '0 (0.0%)'

# TABLE 14.1.1 - DEMOGRAPHICS
rows = [['N'] + [str(N[k]) for k in arms]]

for label, col in [('Age, mean (SD)', 'AGE')]:
    rows.append([label] + [
        f"{a[a.ARM==k][col].mean():.1f} ({a[a.ARM==k][col].std():.1f})"
        for k in arms
    ])

rows.append(['Age, median (min, max)'] + [
    f"{a[a.ARM==k].AGE.median():.1f} ({a[a.ARM==k].AGE.min():.1f}, {a[a.ARM==k].AGE.max():.1f})"
    for k in arms
])

for col, values in [('SEXCD', ['F', 'M']), ('RACECD', sorted(a.RACECD.dropna().unique()))]:
    for v in values:
        rows.append([f'{col}: {v}'] + [
            fmt((a[a.ARM==k][col] == v).sum(), N[k]) for k in arms
        ])

t141 = pd.DataFrame(rows, columns=['Characteristic'] + [f'{k} (N={N[k]})' for k in arms])
t141.to_csv('Table_14_1_1_Demographics.csv', index=False)

# TABLE 14.2.1 - EFFICACY
rows = []
for col in ['SBP_CHG', 'DBP_CHG', 'GLU_CHG']:
    pbo = a[a.ARM=='Placebo'][col].dropna()
    rows.append([col, f'{pbo.mean():.2f} ({pbo.std():.2f})',
                 f'{pbo.mean():.2f}'])

    for arm in arms[1:]:
        x = a[a.ARM==arm][col].dropna()
        diff = x.mean() - pbo.mean()
        t, p = stats.ttest_ind(x, pbo, equal_var=False)
        se = np.sqrt(x.var()/len(x) + pbo.var()/len(pbo))
        df = (x.var()/len(x) + pbo.var()/len(pbo))**2 / (
            (x.var()/len(x))**2/(len(x)-1) + (pbo.var()/len(pbo))**2/(len(pbo)-1)
        )
        margin = stats.t.ppf(0.975, df) * se
        rows.append([f'{col} - {arm}', f'{x.mean():.2f} ({x.std():.2f})',
                     f'{diff:.2f} ({diff-margin:.2f}, {diff+margin:.2f}); p={p:.3f}'])

t142 = pd.DataFrame(rows, columns=['Endpoint / Comparison', 'Mean (SD)', 'Difference vs Placebo (95% CI); p-value'])
t142.to_csv('Table_14_2_1_Efficacy.csv', index=False)

# TABLE 14.3.1 - ADVERSE EVENTS
ae_arm = ae.merge(a[['USUBJID', 'ARM']], on='USUBJID', how='inner')
rows = []

for arm in arms:
    total = N[arm]
    subjects = ae_arm[ae_arm.ARM == arm]
    rows.append([arm, 'Any AE', fmt(subjects.USUBJID.nunique(), total)])

    for sev in subjects['AESEV'].dropna().unique():
        n = subjects.loc[subjects.AESEV == sev, 'USUBJID'].nunique()
        rows.append([arm, f'Severity: {sev}', fmt(n, total)])

    for term in subjects['AETERM'].dropna().unique():
        n = subjects.loc[subjects.AETERM == term, 'USUBJID'].nunique()
        rows.append([arm, f'AE term: {term}', fmt(n, total)])

t143 = pd.DataFrame(rows, columns=['Treatment Arm', 'Adverse Event', 'Subjects n (%)'])
t143.to_csv('Table_14_3_1_AE_Summary.csv', index=False)

# FIGURE 14.2.1 - SBP BOX PLOT
fig, ax = plt.subplots(figsize=(7, 5))
ax.boxplot([a[a.ARM==k].SBP_CHG.dropna() for k in arms],
           tick_labels=[f'{k}\n(N={N[k]})' for k in arms], showmeans=True)
ax.axhline(0, linestyle='--', linewidth=0.8)
ax.set_title('Figure 14.2.1: Systolic BP Change from Baseline by Treatment Arm')
ax.set_ylabel('Change from Baseline in SBP (mmHg)')
ax.set_xlabel('Treatment Arm')
fig.text(0.01, 0.01, 'Source: adam_exp7_analysis.csv (SAFFL=Y)', fontsize=8)
plt.tight_layout()
plt.savefig('Figure_14_2_1.png', dpi=200)
plt.show()

# FIGURE 14.2.2 - FOREST PLOT
fig, ax = plt.subplots(figsize=(7, 5))
labels, diffs, lows, highs = [], [], [], []

for col in ['SBP_CHG', 'DBP_CHG', 'GLU_CHG']:
    pbo = a[a.ARM=='Placebo'][col].dropna()
    for arm in arms[1:]:
        x = a[a.ARM==arm][col].dropna()
        diff = x.mean() - pbo.mean()
        se = np.sqrt(x.var()/len(x) + pbo.var()/len(pbo))
        df = (x.var()/len(x) + pbo.var()/len(pbo))**2 / (
            (x.var()/len(x))**2/(len(x)-1) + (pbo.var()/len(pbo))**2/(len(pbo)-1)
        )
        margin = stats.t.ppf(0.975, df) * se
        labels.append(f'{col}: {arm}')
        diffs.append(diff)
        lows.append(diff-margin)
        highs.append(diff+margin)

y = np.arange(len(labels))
ax.errorbar(diffs, y, xerr=[np.array(diffs)-lows, np.array(highs)-diffs],
            fmt='o', capsize=4)
ax.axvline(0, linestyle='--', linewidth=0.8)
ax.set_yticks(y, labels)
ax.set_xlabel('Mean Difference vs Placebo (95% CI)')
ax.set_title('Figure 14.2.2: Treatment Differences in Clinical Endpoints')
fig.text(0.01, 0.01, 'Source: adam_exp7_analysis.csv (SAFFL=Y)', fontsize=8)
plt.tight_layout()
plt.savefig('Figure_14_2_2.png', dpi=200)
plt.show()

print('\nAll tables and figures created successfully!')
print('Table_14_1_1_Demographics.csv')
print('Table_14_2_1_Efficacy.csv')
print('Table_14_3_1_AE_Summary.csv')
print('Figure_14_2_1.png')
print('Figure_14_2_2.png')
```
